1) com retorno e sem parâmetro:

Crie uma função chamada gerar_senha() que não recebe parâmetros.

Ela retorna uma senha aleatória de 8 caracteres contendo letras, números e caracteres especiais;

Essa senha deve ter letra maiúscula, minúscula, um número e um caractere especial.

Dica: você pode utilizar os a biblioteca random e string.
Exemplos:

import random

maiuscula = random.choice(string.ascii_uppercase)
print(maiuscula)
minuscula = random.choice(string.ascii_lowercase)

numero = random.choice(string.digits)

```python
def criarsenha():
  senha = []
  for i in range(0,3):
    maiuscula = random.choice(string.ascii_uppercase)
    senha.append(maiuscula)

  for i in range(0,2):
     minusculo = random.choice(string.ascii_lowercase)
     senha.append(minusculo)

  for i in range(0,2):
    numero = random.choice(string.digits)
    senha.append(numero)

  for i in range(0,1):
    especial = random.choice(string.punctuation)
    senha.append(especial)

    return(senha)
```

3) Crie uma função com retorno e com parâmetro chamada validar_senha(senha) que recebe uma string e retorna um dicionário contendo:

Chave Significado

tamanho_valido    Verdadeiro se tiver ≥ 8 caracteres
tem_numero  Verdadeiro se possuir número
tem_maiuscula     Verdadeiro se tiver letra maiúscula
tem_minuscula     Verdadeiro se tiver letra minúscula
tem_especial      Verdadeiro se possuir caractere especial
senha_valida      Verdadeiro apenas se todos os critérios forem atendidos

```python
def validar_senha(senha):
    # Inicialmente, todos os critérios começam como False
    tamanho_valido = False
    tem_numero = False
    tem_maiuscula = False
    tem_minuscula = False
    tem_especial = False

    # Verificações
    if len(senha) >= 8:
        tamanho_valido = True

    for char in senha:
        if char.isdigit():
            tem_numero = True
        elif char.isupper():
            tem_maiuscula = True
        elif char.islower():
            tem_minuscula = True
        elif not char.isalnum():  # caractere especial
            tem_especial = True

    # Verifica se todos os critérios são verdadeiros
    if (tamanho_valido and tem_numero and tem_maiuscula
        and tem_minuscula and tem_especial):
        senha_valida = True
    else:
        senha_valida = False

    return {
        "tamanho_valido": tamanho_valido,
        "tem_numero": tem_numero,
        "tem_maiuscula": tem_maiuscula,
        "tem_minuscula": tem_minuscula,
        "tem_especial": tem_especial,
        "senha_valida": senha_valida
    }
```

```python
validar_senha(['R', 'T', 'S', 't', 'w', '6', '4', '>'])
```
